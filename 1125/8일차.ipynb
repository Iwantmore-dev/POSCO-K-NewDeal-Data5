    물을 마신다고 가정해보자.
    1.일어선다
    2.왼쪽으로 3걸음 걸어간다.
    3.멈춘다.
    4.오른쪽으로 3걸음 걷는다.
    5.오른쪽으로 방향 전환
    
    매번 for문을 만들어줄 수 도 없고 설령 했다고 한들 자리가 바뀐다던가 하는 문제 발생시 모든 for문을 고쳐야함. 만약 함수를 사용한다면?

    물마시기 필요시
    '물마시기' 호출->> 함수만 불러주면 된다. 즉, 일련의 동작 묶음을 함수라고 볼 수 있음.
    
    함수는 결국 기능을 만들어주는거다

    중간에 자리가 바뀐다고 생각하면 왼쪽을 오른쪽으로만 바꾼다면
    물마시기 함수가 전부 바뀌게 될것임.

    def 함수(): 로 만든다.


함수란 결국 하나의 기능을 위한 여러개의 명령문을 하나로 묶어서 이름을 붙인것이다.
이름을 부르면 묶어둔 코드덩어리가 나오는것.

내장 함수는 python에서 미리 만들어준 함수고, 예시로는 print, len등이 있음.
built in 함수라고도 부른다.

```python
# user_info 함수를 만들어서 외부에서 이름, 나이, 주소를 입력받으시오
# 입력 받은 자료를 출력하시오
# 함수 외부로 넘기는 값은 20살 보다 작으면 미성년, 이상이면 성인으로 결과를 return하도록 하기

def user_info(name,age, addr):
    print('이름은', name,'나이는',age,'주소는',addr,'입니다.')
    if age >=20 :
        result = '성인입니다.'
    else:
        result = '미성년입니다.'
    
    return result #이때 return에 의해 값이 나오는데 그걸 result에 담았음.


#input으로 외부에서 입력받도록 하기
#현석,19,경기도
name = input('이름을 입력해주세요.: ').strip()
age = input('나이를 입력해주세요.: ').strip()

if age.isnumeric():
    age_int = int(age)
    addr = input('주소를 입력해주세요.: ').strip()
    result = user_info(name, age_int, addr) #값, 혹은 인자 argument
    print(result)
else:
    print('숫자오류입니다.')
```

```python
'''
인사관리 시스템
1.회원추가
입력은 name, age, gender
추가할때는 번호는 자동으로 순차적으로 추가되게 할 것.
memebers=[]  리스트에서 관리한다.

안에 들어갈 자료는 dict로 관리하고 

자료 구성 :  code, name, age, gender로 구성된다. 

code는 자동으로 증가하고  1.2.3 ...

name, age, gender는 직접 입력으로 처리한다. input으로 입력해야할거같음.
'''

name = input('이름을 입력해주세요.: ').strip()
age = input('나이를 입력해주세요.: ').strip()
members=[]
code = 1

if age.isnumeric():
    age_int = int(age)
    gender = input('성별을 입력해주세요. : ').strip()
    members.append({'code':code, 'name':name, 'age':age_int, 'gender':gender})
    code += 1
    print('회원 추가 성공')
    print(members)

else:
    print('숫자오류입니다.')
```

```python
members = []
code = 1

while True: # 참인 경우 while문 실행. break없으면 매번 실행됨.
    name = input('이름을 입력해주세요.: ').strip() #공백 제거
    age = input('나이를 입력해주세요.: ').strip()

    if age.isnumeric(): #age가 숫자라면
        age_int = int(age) #정수형 변환
        gender = input('성별을 입력해주세요.: ').strip()
        #공백 제거하고 성별 마저 받음.

        members.append({'code':code, 'name':name, 'age':age_int, 'gender':gender})
        #code, name, age, gender로 하나의 dict로 만듬.
        #members 자체는 리스트이므로 append로 dict를 하나씩 추가

        code += 1
        #회원 추가 완료되면 code값이 1씩 상승.

        print('회원 추가 성공')
        print(members)

    else: #age가 숫자가 아니라면
        print('숫자오류입니다.')

    answer = input('계속 추가하시겠습니까? y/n: ').strip()
    #숫자 오류가 나고서도 추가 여부를 물어봐서 answer에 저장하기

    if answer == 'n': #n을 눌러서 탈출
        break
```
