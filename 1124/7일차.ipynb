# enumerate : 반복이 가능한 자료를 순회하면서 각 요소의 인덱스와 값을 동시에 가져옴.    

```python
#enumerate
names = ['hong', 'kim','lee']

for data in enumerate(names):
    print(data) # 튜플 전체 출력
    print(data[0],data[1]) # 튜플 인덱싱을 통한 개별 요소 출력
    #print(type(data)) 튜플임.
print('================')

for index, item in enumerate(names):
    print(index, item) #튜플 언패킹.


```

## zip : 여러개의 반복 가능한 자료에서 같은 위치에 있는 요소들을 하나로 묶는것. 리스트의 길이가 다르면 짧은 리스트의 길이에 맞춘다.

```python
#zip
temps = [25,32,28,35,19,31,27]
result = ['이상' if temp >30 else '정상' for temp in temps]

for temp, result in zip(temps, result):
    print(f'{temp}온도는 : {result}입니다.')
```
