```python
#구구단 실습
'''
다음과 같이 구구단을 작성하시오 

2*1=2    3*1=3  4*1=4  5*1=5

....

2*9=18   3*9=27 4*9=36  5*9=45
'''
for j in range(1,10):
    for i in range(2,6):
        print(f'{i}*{j} = {i*j}', end= '    ')
    print('\n')
```

```python
'''
숫자를 입력받아서  해당 구구단을 작성하시오

첫단  2   5   7       aaa  => 숫자오류!

끝단  3   7    5

2~ 3 

5~7 단
'''
num1 = input('첫 번째 숫자를 입력해주세요: ')
num2 = input('두 번째 숫자를 입력해주세요: ')

if num1.isnumeric() and num2.isnumeric():
    num1 = int(num1)
    num2 = int(num2)
    if num1 > num2:
        num1, num2 = num2, num1

    for j in range(1, 10):
        for i in range(num1, num2 + 1):
            print(f'{f"{i}*{j}={i*j}":<12}', end='')
        print()
else:
    print('숫자오류!')
```

```python
'''
단어별 등장 횟수 세기

apple banana apple orange banana apple

결과

apple : 3개

banana : 2개

orange : 1개
'''

text = 'apple banana apple orange banana apple'
#input으로 받고 싶다면
#text = input('단어를 입력해주세요.:')
words = text.split() #text를 스플릿 해서 리스트로 만듬
result = {} #빈 딕셔너리 생성

for word in words:
    if word in result:
        result[word] += 1 #첫 반복의 apple의 경우, word는 result에 없으므로 else에 도달
        #만약 apple이 3,6번째 처럼 반복해서 등장하는 경우에는 dictionary의 value를 하나씩 증가시킴
    else:
        result[word] = 1 #기존에 없었으면 result라는 딕셔너리에 key를 등록하고 value를 1로 만들어줌.
for word in result: #result는 이제 완성된 dict이고, word는 키값만 들고오므로 apple, banana, orange만 존재.
    print(f'{word} : {result[word]}개')# 두번째 {}에서 result[word]는 즉 result['apple'], result['banana'], result['orange']이므로 각 키값에 해당하는 value 들고옴.
```

```python
students = [] #빈 학생 리스트
std_num = input('학생 수를 입력해주세요:')
if std_num.isnumeric(): #만약 input이 숫자면
    std_num = int(std_num) #int형 변환
    
    for i in range(std_num): #std_num의 수만큼 반복해서 학생들의 성,점수 물어볼거임. 예를들어 3명이면 i는 0,1,2
        name = input('학생의 성을 입력해주세요:')
        kor = input('국어 점수를 입력해주세요:')
        eng = input('영어 점수를 입력해주세요:')
        if kor.isnumeric() and eng.isnumeric():
            kor = int(kor)
            eng = int(eng)
            students.append((name, kor, eng)) # 입력한 name, kor, eng 3가지 변수를 하나의 튜플로 묶고, 해당 튜플을 리스트의 요소로 추가함.
            #만약 students list가 주어진다면 불필요함.        
            #그럼 3명이라고 생각하면 students=[(name,kor,eng),(name2,kor2,eng2),(name3,kor3,eng3)]
        else:
            print('숫자오류!')
        #여기서도 input으로 문자를 넣는경우 if문을 이용하여 숫자오류! 표현

    for student in students: #각각의 학생에 대해
        avg = (student[1] + student[2]) / 2 #kor과 eng의 평균
        if avg >= 60: #평균이 60 이상인 경우
            print(f'{student[0]} : 국어 {student[1]}점, 영어 {student[2]}점, 평균 {avg}점')
        else: #평균이 60도 안되면
            print(f'{student[0]} : 국어 {student[1]}점, 영어 {student[2]}점, 평균 {avg}점 평균이 60점 미만입니다!')
else:
    print('숫자오류!')




#평균이  60 이상인 학생들의 자료를 출력하시오
#리스트와 튜플에 관한 문제
#튜플 인덱싱으로 해결함.
```
