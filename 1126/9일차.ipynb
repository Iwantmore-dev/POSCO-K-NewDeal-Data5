```python
#생성자. 값을 하나하나 넣는게 아니라 한번에 넣는거
#생성자는 객체를 만드는 기능임.
'''
생성자는 객체를 생성한다.
객체가 만들어질때 자동으로 호출
객체가 만들어질때 해야 할 일들을 기술하거나
객체안의 변수 (인스턴스 변수) 초기화할때 사용한다.

'''
class person():
    def __init__(self, name, age, gender, addr):
        self.name = name #self.name은 인스턴스 변수
        self.age = age
        self.gender = gender
        self.addr = addr
    #클래스 내부에서 현재 작업하고 있는 객체가 무엇인지 구분하기 위해 self를 사용한다는 의미
p1 = person('lee',27,'남', '서울')
print(p1.name, p1.age, p1.gender, p1.addr)
```

```python
class student:
    def __init__(self, name:str, kor:int, eng:int, math:int):
        self.name = name
        self.kor = kor
        self.eng = eng
        self.math = math
    def prt(self):
        print('이름:',self.name)
        print('국어:',self.kor)
        print('영어:',self.eng)
        print('수학:',self.math)

        tot = self.kor+self.eng+self.math
        print('총점',tot)

        avg = tot/3
        print('평균',avg)
```

```python
s1 = student('hong',90,50,70)
s2 = student('lee',70,50,100)
s3 = student('yoon',90,70,80)
s1.prt()
```

```python
class NewStudent:
  def __init__(self, name:str, kor:int=0,eng:int=0, math:int=0):
     self.name=name
     self.kor=kor
     self.eng=eng
     self.math=math

  def get_total(self):
    tot=self.kor+self.eng+self.math
    return tot

  def get_avg(self, tot):
     return tot/3

  def prt(self):
    print('이름:',self.name)
    print('국어:', self.kor)
    print('영어:', self.eng)
    print('수학:', self.math)

    result=self.get_total()
    print('총점:', result)

    avg=self.get_avg(result)
    print('평균', avg)
```

```python
new_s1=NewStudent('hong',100,90,80)
new_s1.prt()
```

```python
#주민번호 101010 - 3010111
#출력은
#생년월일
#성별
#나이

#입력을 받아서 생년월일, 성별, 나이 출력해보기
# 주민등록번호 입력

import datetime

def person_info(numb):
    year = datetime.datetime.now().year
    front, back = numb.split('-')
   
    
    birth = int(front[:2]) # [01 23 45] 중에서 0번, 1번 인덱스가 출생년도
    month = int(front[2:4]) # 출생 달
    day = int(front[4:6]) # 출생 일

    # 출생 연도 계산
    if back.startswith('1') or back.startswith('2'):
        birth = birth + 1900
    elif back.startswith('3') or back.startswith('4'):
        birth = birth + 2000 #2000년생 이후부터는 3,4로 나오므로
    
    if back.startswith('1') or back.startswith('3'):
        gender = '남성'
    else: #2 or 4
        gender = '여성'

    # 연 나이 계산
    age = year - birth

    return birth, month, day, gender, age

name =input('이름을 입력해주세요:')
numb = input('주민번호를 입력해주세요, (-)포함.:') #101010-1111111 이면 필요한건 앞자리 6자리와 뒷자리 1자리 필요


result = person_info(numb) #result에는 birth,month, day, gender, age 값이 담김

print('이름:', name)
print('생년월일:', result[0], result[1], result[2])
print('성별:', result[3])
print('나이:', result[4])
```
