```python
#else와 finally 코드

text = "62.5"
try:
    temp=float(text)
    print(temp)
except ValueError as e:
    print('변환 실패', e)
else:
    print(temp) #리더 피셜 별로 안 깔끔하다. 굳이 안써도 된다.
finally:
    print('종료')
```
